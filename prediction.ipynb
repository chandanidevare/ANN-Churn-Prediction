```python
import tensorflow as tf
from tensorflow.keras.models import load_model
import pickle
import pandas as pd
import numpy as np
```

```python
# Load the trained model, scaler pickle file , onehot
model=load_model('model.h5')

# Load the encoder & scaler pickle file:
with open('onehot_encoder_geo.pkl', 'rb') as file:
    onehot_encoder_geo=pickle.load(file)

with open('label_encoder_gender.pkl', 'rb') as file:
    label_encoder_gender=pickle.load(file)

with open('scaler.pkl', 'rb') as file:
    scaler=pickle.load(file)
```

```python
# Example input data for prediction:
input_data={
    'CreditScore':700,
    'Geography':'France',
    'Gender':'Male',
    'Age':30,
    'Tenure':3,
    'Balance':80000,
    'NumOfProducts':2,
    'HasCrCard':1,
    'IsActiveMember':1,
    'EstimatedSalary':50000
}
```

```python
# Onehot encoded the 'Geography' columns:
geo_encoded= onehot_encoder_geo.transform([[input_data['Geography']]]).toarray()

geo_encoded_df= pd.DataFrame(geo_encoded,columns=onehot_encoder_geo.get_feature_names_out(['Geography']))
geo_encoded_df
```

```python
input_df= pd.DataFrame([input_data])
input_df
```

```python
# Concatenation onehot encoded:
input_df= pd.concat([input_df.drop(['Geography'],axis=1),geo_encoded_df],axis=1)
input_df
```

```python
# Encode the Categorical Variables:
input_df['Gender']= label_encoder_gender.fit_transform(input_df['Gender'])
input_df
```

```python
# Scaling the input data:
input_scaled = scaler.transform(input_df)
```

```python
input_scaled=scaler.transform(input_df)
input_scaled
```

```python
# Prediction the Chrun Probability:
churn_probability=model.predict(input_scaled)
churn_probability
```

```python
churn_proba=churn_probability[0][0]
churn_proba
```

```python
if churn_proba > 0.5:
    print("The Customer is likely to churn.")
else:
    print("The Customer is not likely to churn.")
```
